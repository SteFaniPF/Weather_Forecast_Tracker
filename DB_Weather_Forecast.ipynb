```python
# https://docs.google.com/spreadsheets/d/14K9w_xPU3MoS-GbuXlmdfZhqSzS0-UrgonsR80RximI/edit?usp=sharing



import pandas as pd
import requests
import time

#Alterar o fim para formato CSV = add:  /expor?format=csv
endpoint_planilha = "https://docs.google.com/spreadsheets/d/14K9w_xPU3MoS-GbuXlmdfZhqSzS0-UrgonsR80RximI/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
print ('---Dados Coletados---')
dados_planilha = pd.read_csv(endpoint_planilha)

display(dados_planilha)
```

```python
#iterar cada resposta do formulario


#iterrows le uma linha por vez numa base de dados
for indice, linha in dados_planilha.iterrows():
  nome_local = linha['Nome do Local']
  lat = linha['Latitude']
  lon = linha['Longetude']

  parametros = {
      'latitude' : lat,
      'longitude' : lon,
      'current': "temperature_2m"
  }
  print(f'-----{nome_local}-----')
  resposta = requests.get(endpoint_api, params=parametros).json()

  display(resposta)

  print('\n\n')
```
